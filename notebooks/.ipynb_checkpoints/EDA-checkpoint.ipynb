```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import eda_helper_functions
from sklearn.ensemble import IsolationForest
```

```python
project_dir=r"C:\Users\koush\anaconda3\flights-sagemaker"
data_dir="data"
```

```python
train=pd.read_csv(os.path.join(project_dir,data_dir,'train.csv'))
```

```python
train
```

```python
train.dtypes
```

```python
train=train.assign(**{
    col: pd.to_datetime(train.loc[:,col],dayfirst=True,format="mixed")
    for col in ["date_of_journey","dep_time","arrival_time"]
})
```

#### High level analysis of Outliers

```python
forest=IsolationForest(n_estimators=10,random_state=42)
train.assign(outlier=forest.fit_predict(train.drop(columns='price').select_dtypes(include='number'))).query("outlier==-1")
```

#### Date of Journey

```python
train.date_of_journey
```

```python
eda_helper_functions.dt_univar_plots(data=train,var='date_of_journey',target='price')
```

```python
train.groupby(pd.Grouper(key='date_of_journey',freq='W')).price.mean().to_frame().rename_axis(index='weekly').plot(kind='bar',figsize=(8,4),edgecolor='black')#.set_axis(['March','April','May','June'],axis=0)


ax=plt.gca()
eda_helper_functions.rotate_xlabels(ax)
```

```python
train.groupby([pd.Grouper(key='date_of_journey',freq='M'),'source']).price.mean().unstack(fill_value=0).set_axis(['March','April','May','June'],axis=0).rename_axis(index='month').plot(kind='bar',figsize=(12,4))

ax=plt.gca()
ax.legend(
loc="upper right",bbox_to_anchor=(1,1))
eda_helper_functions.rotate_xlabels(ax)
```
